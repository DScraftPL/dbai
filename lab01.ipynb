```python
import seaborn as sns

df = sns.load_dataset("penguins")

```

```python
df.nlargest(n=5, columns=["body_mass_g"])
```

```python
df.groupby('species').agg(n=("body_mass_g", "size"), n_mass=(
    "body_mass_g", "count"), mean_mass=("body_mass_g", "mean")).sort_values("mean_mass")
```

```python
df.groupby(["species", "island"]).agg(n=("body_mass_g", "size"), mean_mass=("body_mass_g", "mean"))
```

```python
df.groupby(["species", "island"]).agg(
    n=("body_mass_g", "size"), mean_mass=("body_mass_g", "mean")
).reset_index()
```

```python
summary = df.groupby(["species", "island"]).agg(n=("body_mass_g", "size"), mean_mass=("body_mass_g", "mean")).reset_index()

summary[summary["n"] >= 50]
```

# ZADANIA SAMODZIELNE

```python
df = sns.load_dataset("tips")

df.info()
```

```python
df.isna().mean()
```

```python
print(df["sex"].value_counts())
print(df["smoker"].value_counts())
print(df["day"].value_counts())
print(df["time"].value_counts())
```

```python
df.groupby('day').agg(mean_total_bill=("total_bill", "mean")).sort_values(by=["mean_total_bill"], ascending=False)
```

```python
df['bill_per_person'] = df['total_bill']/df['size']

result = df.groupby('size').agg(bill=('bill_per_person', "mean")).reset_index()

print(result.info())
print(result.head())

result[(result['size'] > 4) | (result['size'] < 3)]
```

```python
df[['total_bill', 'tip', 'day', 'size']].sort_values(by=['total_bill'], ascending=False).head(n=10)
```

```python
df.groupby('day').agg(mean_bill = ('total_bill', 'mean'), count_bill = ('total_bill', 'count'), mean_tip = ('tip', 'mean')).sort_values(by="mean_bill", ascending=False)
```

```python
summary = df.groupby("day").agg(mean_bill=("total_bill", "mean"))

summary[summary["mean_bill"] > 20]
```
