```python
df = sns.load_dataset("planets")
dirty=df.copy()
dirty.loc[15, "method"] = "radial velocity" 
dirty.loc[25, "distance"] = 99999.0 
dirty.loc[35, "orbital_period"] = -1.0 
dirty.loc[45, "number"] = 0 
dirty = pd.concat([dirty, dirty.iloc[[1,2]]],ignore_index=True)

print(f"Braki:\n{dirty.isna().sum()}")
print(f"Duplikaty: {dirty.duplicated().sum()}")
print(f"Unikalne wartości method:\n{dirty["method"].value_counts()}")
print(f"Zakresy wartości:\n {dirty.describe().loc[["mean","std","min","max"]].round(2)}")
```

```python
import pandera.pandas as pa
from pandera.pandas import Check, Column, DataFrameSchema

METHOD_COLUMNS="Pulsation Timing Variations", "Pulsar Timing", "Astrometry", "Orbital Brightness Modulation", "Transit Timing Variations", "Eclipse Timing Variations", "Microlensing", "Imaging", "Transit" , "Radial Velocity"

schemat = DataFrameSchema({
 "method": Column(str, Check.isin(METHOD_COLUMNS)),
 "number": Column(int, Check.in_range(0, 8), nullable=True),
 "orbital_period": Column(float, Check.in_range(0, 730050), nullable=True),
 "mass": Column(float, Check.in_range(0, 25), nullable=True),
 "distance": Column(float, Check.in_range(0, 8550), nullable=True),
 "year": Column(int, Check.in_range(1987, 2027), nullable=True)})

validated = schemat.validate(df) 
print("Walidacja zakończona powodzeniem, wymiary:", validated.shape) 

try:
  schemat.validate(dirty, lazy=True)
  print("Walidacja zakończona powodzeniem")
except pa.errors.SchemaErrors as err:
  print(err.failure_cases)
```

```python
dirty.info()

dirty = dirty.drop_duplicates()
dirty.value_counts('method')
dirty = dirty[dirty['method'] != 'radial velocity']
dirty.value_counts('method')

dirty = dirty[dirty['number'] > 0]
dirty = dirty[dirty['distance'] < 9000]
dirty = dirty[dirty['orbital_period'] > 0]
dirty.describe()

print(dirty.isna().sum())
dirty = dirty.dropna()
print(dirty.isna().sum())

print(dirty.shape)

try:
  schemat.validate(dirty, lazy=True)
  print("Walidacja zakończona powodzeniem")
except pa.errors.SchemaErrors as err:
  print(err.failure_cases)
```

```python
from sklearn.model_selection import train_test_split

X = dirty[['number', 'orbital_period', 'mass', 'distance', 'method']]
y = dirty['year'] >= 2010

X_tre, X_test, Y_tre, Y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print(X_tre.shape)
print(X_test.shape)

print(Y_tre.shape)
print(Y_test.shape)

print(Y_tre.value_counts())
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_columns = ["distance", "mass", "orbital_period", "number"]
categorical_columns = ['method']

numeric_pipeline = Pipeline([
 ("imputer", SimpleImputer(strategy="median")),
 ("scaler", StandardScaler()), 
])
categorical_pipeline = Pipeline([
 ("imputer", SimpleImputer(strategy="most_frequent")), 
 ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)), 
])

preprocessor = ColumnTransformer([
 ("numeric", numeric_pipeline, numeric_columns),
 ("categorical", categorical_pipeline, categorical_columns),
])

X_train_processed = preprocessor.fit_transform(X_tre)
X_test_processed = preprocessor.transform(X_test)

print(X_train_processed.shape)
print(X_test_processed.shape)

print(pd.DataFrame(X_train_processed).describe())
print(preprocessor.get_feature_names_out())

## TO CLAUDE! 

# Mediany nauczone przez imputer (kolejność = numeric_columns)
num_imputer = preprocessor.named_transformers_["numeric"].named_steps["imputer"]

imputer_medians = pd.Series(num_imputer.statistics_, index=numeric_columns)
print(imputer_medians)

median_imputer = imputer_medians["distance"]
median_train   = X_tre["distance"].median()    # pandas pomija NaN, tak jak SimpleImputer
median_full    = dirty["distance"].median()   # podmień na nazwę swojego oczyszczonego DataFrame'u

comparison = pd.DataFrame({
    "źródło": ["imputer (pipeline)", "X_train", "cały oczyszczony zbiór"],
    "mediana distance": [median_imputer, median_train, median_full],
})
print(comparison)

print("imputer == X_train:", np.isclose(median_imputer, median_train))
print("różnica względem całego zbioru:", median_full - median_imputer)
```
